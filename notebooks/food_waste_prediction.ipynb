# Food Waste Prediction Notebook
This notebook visualizes cafeteria waste and explores trends.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../data/food_waste_records.csv')
```

```python
df['Waste_Rate'] = df['Waste_kg'] / df['Meals_Prepared']
df.groupby('Cafeteria')['Waste_Rate'].plot(legend=True)
plt.title('Daily Waste Rate by Cafeteria')
plt.ylabel('Waste per Meal Prepared')
plt.xlabel('Day Index')
plt.show()
```
